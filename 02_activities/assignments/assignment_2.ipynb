#### Starter Code for Question 2

```python
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val = 0, left = None, right = None):
#         self.val = val
#         self.left = left
#         self.right = right
```

```python
#bt_path(root: TreeNode) -> List[List[int]]:

def bt_path(root, lvl=0, current_path=None):
    if current_path == None:
        current_path = []
    
    tree_len = len(root)

    # Base case: Index out of bounds or node is missing
    if lvl >= tree_len or root[lvl] == None:
        return []

    # Add current node to the path
    current_path.append(root[lvl])
    
    left_child_index = 2 * lvl + 1
    right_child_index = 2 * lvl + 2
    
    # Check if this node is a leaf
    is_left_empty = left_child_index >= tree_len or root[left_child_index] == None
    is_right_empty = right_child_index >= tree_len or root[right_child_index] == None
    
    if is_left_empty and is_right_empty:
        
        result = [list(current_path)]
        
    else:
        
        result = bt_path(root, left_child_index, current_path) + \
                 bt_path(root, right_child_index, current_path)
    
    current_path.pop() #remove the current node before backtracking

    return result
```

```python
from typing import List

def first_duplicate(nums: List[int]) -> int:
    seen = set()
    for num in nums:
        if num in seen:
            return num
        seen.add(num)
    return -1
```
